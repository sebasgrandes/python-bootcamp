```python

# otra forma de abrirlo sin necesidad de cerrarlo luego...
# por defecto esta en mode:"r" (de read)
with open("myfilew.txt") as my_new_file:
    contents = my_new_file.read()

print(contents) # se lee sin problemas


```

```python

# existen los modos "r", "w", a de append, r+ de reading y writing, w+ de writing y reading pero sobreescribir y crear un nuevo archivo creo.
with open("myfilew.txt", mode="a") as f:
    f.write("\n asdasd \n qweqwe \n texto agregado hay que probarlo")
    
with open("myfilew.txt") as my_new_file:
    contents2 = my_new_file.read()
print(contents2)

```

```python

# si no lo encuentra... lo escribe
with open("asdasdasd.txt", mode="w") as f:
    f.write("I CREATED THIS FILE")

with open("asdasdasd.txt", mode="r") as qwe:
    print(qwe.read())
```
