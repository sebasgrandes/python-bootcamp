```python

# también se puede iterar en una tupla... pero lo mas curioso es el tuple packing

my_lista = [(1,2,"b"),(2,3,"c"),(4,5,"d"),(6,7,"e")]
# puedes poner el tuple packing con o sin parentesis
for a,b,c in my_lista:
    print(f"primer elemento: {a}... tercer elemento: {c}")

# también se puede iterar en diccionarios como si nada... pero de nuevo, lo interesante esta en lo siguiente:
d = {'k1': 1, 'k2': 2, 'k3': 3}
# también podrias usar d.values() o simplemente d.keys() o sino d nada mas y ya
for a,b in d.items():
    print(a,b)
    

```
