```python

# también se puede iterar en una tupla... pero lo mas curioso es el tuple packing

my_lista = [(1,2,"b"),(2,3,"c"),(4,5,"d"),(6,7,"e")]
# puedes poner el tuple packing con o sin parentesis
for a,b,c in my_lista:
    print(f"primer elemento: {a}... tercer elemento: {c}")

# también se puede iterar en diccionarios como si nada... pero de nuevo, lo interesante esta en lo siguiente:
d = {'k1': 1, 'k2': 2, 'k3': 3}
# también podrias usar d.values() o simplemente d.keys() o sino d nada mas y ya
for a,b in d.items():
    print(a,b)
    

```

```python

# - 36 while loops

x = 0

while x < 5:
    print("helou")
    x += 1
else:
    print("fin")


z = [1,2,3]

# pass se usa para evitar sintax errors y lo demas (supuestamente no hace nada xd)

for item in z:
    pass
print("el script fue pass")

# continue: va a la parte superior del cierre del loop mas cercano
for letter in "Sebastian":
    if letter =="a":
        continue
    print(letter)

# break: para el loop
for letter in "Sebastian":
    if letter =="a":
        break
    print(letter)


```

```python

# - 37 useful operators

# range

# también puedes colocarle range(11) y listo, empezará desde el 0 hasta el 10
for number in range(2,11,2):
    print(number)
    
    
# range es un generador, genera numeros digamos pero no se almacenan en memoria, por eso al imprimir no sale nada... pero si puedes transformarlo a list e imprimirlo
print(range(0,10))
print(list(range(0,10)))


# . enumerate: toma una item y lo convierte en tupla digamos... siendo el 1er elemento el indice y el 2do el valor o el item en sí mismo
# + el enumerate es muy util porque de esa forma no tienes que crear una variable que simule el indice y que se vaya sumando en cada iteracion... python le da una solucion a esto mediante el enumerate

word = "Sebas"
for w in enumerate(word):
    print(w) # las tuplas como tal
for index, value in enumerate(word):
    print(f"el indice es {index} y el valor es {value}") # las tuplas como tal


```

```python

# zip: supuestamente es lo inverso a enumerate... toma varias listas y las coloca como tuplas
list1 = [1,2,3,4,5]
list2 = ["a", "b", "c"]
list3 = [100, 200, 300]

# . también puedes en lugar de item poner... a,b,c... o solo "b" (a esto se le llama desenpacar)
for item in zip(list1, list2, list3):
    print(item)

print(list(zip(list1,list2,list3)))

```
